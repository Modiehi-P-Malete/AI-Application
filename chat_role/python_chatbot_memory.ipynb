OBJECTIVE:
Build a chatbot that:

- introduces itself
- answers a Python question
- summarizes its own previous answer

💡 Hint:

- use a `messages` list
- append both user and assistant messages

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
load_dotenv(override=True)

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
system_message = """
You are a helpful Python tutor chatbot.
Introduce yourself politely.
Answer Python questions clearly.
If asked to summarize, summarize your previous response.
Respond in markdown.
"""
```

```python
messages = [
    {
        "role": "system",
        "content": system_message
    }
]

def python_chatbot(user_input):

    # Store user message
    messages.append({
        "role": "user",
        "content": user_input
    })

    # Send conversation history to the model
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=messages
    )

    # Extract assistant's reply
    assistant_reply = response.choices[0].message.content

    # Store assistant reply
    messages.append({
        "role": "assistant",
        "content": assistant_reply
    })

    return assistant_reply
```

```python
view = gr.Interface(
    fn=python_chatbot,
    inputs=gr.Textbox(
        label="Ask a Python Question",
        lines=5,
        placeholder="Example: What is a Python dictionary?"
    ),
    outputs=gr.Markdown(label="Chatbot Response"),
    title="Python Tutor Chatbot",
    description="Ask Python questions and request summaries.",
    flagging_mode="never"
)

view.launch()
```
