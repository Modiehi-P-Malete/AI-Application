```python

import gradio as gr
```

```python
def greet(text):
    print(f"Shout has been called with input: {text}")
    return text.upper()

greet("Hey there!")
```

```python
simple_chatbot = gr.Interface(fn=greet, inputs="text", outputs="text", flagging_mode="never")
```

```python
simple_chatbot.launch(inbrowser=True, auth=("patience", "mellow123"))
```
